Import Library

```python
import matplotlib.pyplot as plt
```

line plot

```python

x=[1,2,3,4,5]
y=[10,20,30,40,50]

plt.plot(x,y, color='red',marker='s', linestyle="--" ,linewidth= 2)
plt.xlabel('X-axis')
plt.ylabel('y-axis')
plt.title('Line Plot')
plt.show()
```

Scatter plot

```python
x=[1,2,3,4,5]
y=[10,20,30,40,50]

plt.scatter(x,y, color='red')
plt.xlabel('X-axis')
plt.ylabel('y-axis')
plt.title('Scatter Plot')
plt.show()
```

3. Histogram

```python
marks = [45,46,47,48,50,55,60,70,85,90]


plt.hist( marks, bins=5,edgecolor='black')
plt.show()
```

4. Bar Chart

```python
subject= ["Bangla", 'English', 'Math', 'Science','ICT']

marks=[75,55,45,68,70]


plt.bar(subject,marks)
plt.show()
```

5. Pichart

```python
subject= ["Bangla", 'English', 'Math', 'Science','ICT']

marks=[40,20,15,20,5]


plt.pie(marks, labels=subject, autopct='%1.1f%%')
plt.show()
```

**Seaborn**

1. line plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [10,20,15,25,30]


sns.lineplot(x=x, y=y)
plt.show()
```

scatter plot

```python
x = [1,2,3,4,5]
y = [10,20,15,25,30]


sns.scatterplot(x=x, y=y)
plt.show()
```

3 histogram

```python
marks = [45,46,47,48,50,55,60,70,85,90]


sns.histplot( marks, bins=5,edgecolor='black')
plt.show()
```

Bar chart
